```python
sns.histplot(df['AMT_CREDIT'])
```

## AMT_CREDIT Distribution

- Right-skewed distribution. Majority of loans fall between 0.2 and 1.0 (scaled units).
- A small number of extremely high credit amounts exist — these are outliers.
- No capping applied here. Binning in the next stage will naturally
  group these extremes into the top bin, so they will not distort the model.

```python
sns.boxplot(x='TARGET', y='AMT_CREDIT', data=df)
```

```python
sns.histplot(df['AMT_ANNUITY'])
```

## AMT_ANNUITY Distribution

- Right-skewed. Majority of annuity amounts fall below 50,000.
- Extreme outliers present beyond 150,000 — very few cases.
- No capping applied. Binning will absorb these into the top bin downstream.

```python
sns.boxplot(x='TARGET', y='AMT_ANNUITY', data=df)
```
